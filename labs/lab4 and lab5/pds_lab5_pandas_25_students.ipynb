Series study

```python
# dunder method used to check the pandas version
print("Pandas version:", pd.__version__)

# Test creating a serie
series_test = pd.Series([3,'string value',5,7])
# It converts the series values 
# You can see that if the series has objects of a diff type it will convert the typing of the objects to object like dtype: object
print(series_test)

# You can give labels to the series , if you dont the labels will be the indexes
serie_labels = ['label for the 1st row', 'label for the 2nd row']
series_with_labels = pd.Series([1, 3], index=serie_labels)
# here since the types are all the same pandas will convert them to 'dtype: int64'
print(series_with_labels)

# You can use the series to perform operations by label, one example would be to sum the total of each type of label 
# The series automatically will add the values with the same label
same_labels = ['skoda', 'BMW']
diff_labels = ['skoda', 'volkswagen']

# This series are created with the same labels it will be straightforward
series1 = pd.Series([123, 68], index=same_labels)
series2 = pd.Series([524, 6], index=same_labels)

# Operations perform for all rows at label level
print("Operation with series with the same indexes")
print(series1 + series2)
print(series1 * 0)

# Series created with diff labels
series_diff_labels = pd.Series([32, 3], index=diff_labels)

# Check how Series with diff labels perform operations
print("Operation with series with diff indexes")
# Notice that for the indexes that do not exist in both the series they appear Nan
print(series1 + series_diff_labels)

# Diff between map and apply
# map applies a function elementwise
# apply applies a function to the entire series
series_copy = series1.copy() 
series_copy.apply(lambda x: x * 2)
print("Before apply")
print(series1)
print("After apply")
print(series_copy)

# Filter series by condition 
series_filtered = series1[series1 > 100]
print("Non filtered series: ")
print(series1)
print("Filtered Series: ")
print(series_filtered)

# Series behaves like a dict 
# We can initialize a series like a dict for example 
dict = {"skoda" : 1 , 
        "ferrari" : 2}

series_initialized_with_dict = pd.Series(dict)
print("Series initialized with a dictionary")
print(series_initialized_with_dict)

# Or using another series with other indexes 
other_index_list = ["aston_martin", "skoda", "mercedes"]
series = pd.Series(series_initialized_with_dict, index=other_index_list)
print("Series initialized with another serie")
print(series)

# Using Series methods: isnull(), notnull(), fillna(0), astype(int)
print("\nCalling Series methods:")

print("\nseries.isnull():")
print(series.isnull())

print("\nseries.notnull():")
print(series.notnull())

print("\nseries.fillna(0):")
print(series.fillna(0))

print("\nseries.fillna(0).astype(int):")
print(series.fillna(0).astype(int))

print(series.index)
print(series.name)
```
